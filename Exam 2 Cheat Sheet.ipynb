# Definitions

$$
e^{i\theta} = \cos\theta + i\sin\theta
$$
$$
e^{-i\theta} = \cos\theta - i\sin
$$

Trigonometric functions defined via exponentials
$$
\cos z = \frac{e^{iz} + e^{-iz}}{2}
$$
$$
\sin z = \frac{e^{iz} - e^{-iz}}{2i}
$$

Hyperbolic functions defined via exponentials
$$
\cosh z = \frac{e^z + e^{-z}}{2}
$$
$$
\sinh z = \frac{e^z - e^{-z}}{2}
$$


# Rectangular decomposition

Trigonometric functions with complex arguments
$$
\cos(x+iy) = \cos x \cosh y - i \sin x \sinh y
$$
$$
\sin(x+iy) = \sin x \cosh y + i \cos x \sinh y
$$

Hyperbolic functions with complex arguments
$$
\cosh(x+iy) = \cosh x \cos y + i \sinh x \sin y
$$
$$
\sinh(x+iy) = \sinh x \cos y + i \cosh x \sin y
$$

# Inverses

Inverse trigonometric functions
$$
\arcsin z = -i \ln\Big( i z + \sqrt{1-z^2} \Big)
$$
$$
\arccos z = -i \ln\Big( z + i\sqrt{1-z^2} \Big)
$$
$$
\arctan z = \frac{i}{2} \ln\Big( \frac{1-i z}{1+i z} \Big)
$$

Inverse hyperbolic functions
$$
\operatorname{arcsinh} z = \ln\Big( z + \sqrt{z^2+1} \Big)
$$
$$
\operatorname{arccosh} z = \ln\Big( z + \sqrt{z^2-1} \Big)
$$
$$
\operatorname{arctanh} z = \frac{1}{2} \ln\Big( \frac{1+z}{1-z} \Big)
$$

# Roots and Periods



Roots (zeros) and Periodicity for Trigonometric Functions

$$
\cos z = 0 \quad \text{if} \quad z = \frac{\pi}{2} + \pi k, \quad k \in \mathbb{Z}
$$
$$
\sin z = 0 \quad \text{if} \quad z = \pi k, \quad k \in \mathbb{Z}
$$
$$
\cos(z+2\pi) = \cos z,\quad \sin(z+2\pi) = \sin z
$$

Roots (zeros) and Periodicity for Hyperbolic Functions

$$
\cosh z = 0 \quad \text{if} \quad z = i\pi\left(k+\frac{1}{2}\right), \quad k \in \mathbb{Z}
$$
$$
\sinh z = 0 \quad \text{if} \quad z = i\pi k, \quad k \in \mathbb{Z}
$$
$$
\cosh(z+2\pi i) = \cosh z,\quad \sinh(z+2\pi i) = \sinh z
$$


# Linear Transforms

$$
\cosh(a+b) = \cosh a \cosh b + \sinh a \sinh b
$$
$$
\sinh(a+b) = \sinh a \cosh b + \cosh a \sinh b
$$


Addition formulas:
$$
\sin(a+b)=\sin a \cos b + \cos a \sin b
$$
$$
\cos(a+b)=\cos a \cos b - \sin a \sin b
$$

Subtraction formulas:
$$
\sin(a-b)=\sin a \cos b - \cos a \sin b
$$
$$
\cos(a-b)=\cos a \cos b + \sin a \sin b
$$

Product-to-sum formulas:
$$
\sin a \sin b = \frac{1}{2}\Big[\cos(a-b)-\cos(a+b)\Big]
$$
$$
\cos a \cos b = \frac{1}{2}\Big[\cos(a+b)+\cos(a-b)\Big]
$$
$$
\sin a \cos b = \frac{1}{2}\Big[\sin(a+b)+\sin(a-b)\Big]
$$

Sum-to-product formulas:
$$
\sin a + \sin b = 2\sin\left(\frac{a+b}{2}\right)\cos\left(\frac{a-b}{2}\right)
$$
$$
\sin a - \sin b = 2\cos\left(\frac{a+b}{2}\right)\sin\left(\frac{a-b}{2}\right)
$$
$$
\cos a + \cos b = 2\cos\left(\frac{a+b}{2}\right)\cos\left(\frac{a-b}{2}\right)
$$
$$
\cos a - \cos b = -2\sin\left(\frac{a+b}{2}\right)\sin\left(\frac{a-b}{2}\right)
$$

# Assumptions for Contour Integration in Complex Analysis

1. **Piecewise Smooth Contour**
   - The contour \( C \) must be piecewise smooth so that \( dz \) is well-defined.

2. **Function Defined on and Near Contour**
   - The function \( f(z) \) must be defined on the path and in its vicinity.

3. **Analytic Function (for Cauchy's Theorem)**
   - \( f(z) \) is **analytic** on and inside \( C \):
     $$
     f'(z) \text{ exists and is continuous}
     $$

4. **Isolated Singularities (for Residue Theorem)**
   - Any singularities inside \( C \) must be isolated and finite in number.

5. **Closed Contour (for many theorems)**
   - The path \( C \) should start and end at the same point:
     $$
     \oint_C f(z)\, dz
     $$

6. **Proper Orientation**
   - Counterclockwise is the positive orientation. Clockwise reverses the sign.

7. **Correct Parametrization**
   - If using a parameter like \( \theta \):
     $$
     z = z(\theta), \quad dz = \frac{dz}{d\theta} d\theta
     $$

8. **Convergence (for infinite or improper integrals)**
   - Ensure the integral converges if the path is unbounded or near singularities.
