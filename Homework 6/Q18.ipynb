**4. Let $C$ be any simple closed contour, described in the positive sense in the $z$ plane and write**

$$
g(z) = \int_C {s^3 + 2s \over (s - z)^3} ds
$$

**Show that $g(z)= 6πiz$ when $z$ is inside $C$ and that $g(z)= 0$ when $z$ is outside.**

The numerator $f(s) = s^3 + 2s$ is analytic on $\mathbb{C}$. So, when $z$ is interior to the contour, we can use the Cauchy integral theorem with $n = 2$ to write,

$$
g(z) = {2i\pi \over 2!} f''(z)
$$

Find the second derivative.

$$
f'' = 6s
$$

So,

$$
g(z) = {2i\pi \over 2} 6(z) = 6i\pi z
$$

When z is outside the contour, then $g(z)$ is analytic with respect to $s$ on the contour and its interior points. It's the ratio of two nonzero polynomials. So, the the path integral along a closed contour must be zero.

Note that we have not addressed the case when $z$ is on the contour.
