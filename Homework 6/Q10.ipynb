**Let C be the arc of the circle $|z| = 2$ from $z = 2$ to $z = 2i$**

**1. Without evaluating the integral, show that**

**(a)**
$$
\left|
    \int_C {z - 4 \over z^3 - 1}dz
\right| \le
{6\pi \over 7}
$$

The radius of the circle is $4\pi$, so the length of the contour $C$ is $\pi$.

Note that,

$$
|z - 4| \le |z| + 4 = 6
$$

$$
|z^3 - 1| \ge ||z^3| - 1| = ||z|^3 - 1| = 7
$$

So,

$$
\left| {z - 4 \over z^3 - 1} \right| \le {6 \over 7}
$$

So,

$$
\left| \int_C {z - 4 \over z^3 - 1} dz \right| \le {6\pi \over 7}
$$

**(b)**

$$
\left|
    \int_C {1 \over z^2 - 1}dz
\right| \le
{\pi \over 3}
$$

Note that,

$$
|z^2 - 1| \ge ||z^2| - 1| = ||z|^2 - 1| = 3
$$

So,

$$
\left|{1 \over z^2 - 1}\right| \le {1 \over 3}
$$

So,

$$
\left|\int_C {1 \over z^2 - 1} dz \right| \le {\pi \over 3}
$$
