**7. Let $C$ be the unit circle $z = e^{iθ}$ $(−π ≤ θ ≤ π)$. First show that for any real constant $a$,**

$$
\int_C {e^{az} \over z}dz = 2\pi i
$$

**Then write this integral in terms of $θ$ to derive the integration formula**

$$
\int_0^\pi \exp(a \cos (\theta))\cos(a \sin(\theta))d\theta = \pi
$$

The integral can be written,

$$
\int_C {e^{az} \over z}dz = \int_C {e^{az} \over z - 0}dz
$$

Since $0$ is interior to the contour, and the function $f(z) = e^{az}$ is analytic on $\mathbb{C}$, we can use the Cauchy integral theorem to rewrite,

$$
\int_C {e^{az} \over z}dz = 2\pi i f(0) = 2\pi i
$$

We will perform the substitution,

$$
z = \exp(i\theta)
$$

So,

$$
\int_C {e^{az} \over z}dz = \int_{-\pi}^\pi {\exp(a\exp(i\theta)) \over \exp(i\theta)} (i\exp(i\theta)d\theta)
$$

So,

$$
\int_{-\pi}^\pi \exp(a\exp(i\theta))d\theta = 2\pi
$$

Apply Euler's identity,

$$
\int_{-\pi}^\pi \exp(a(\cos(\theta) + i\sin(\theta)))d\theta = 2\pi
$$

Then,

$$
\int_{-\pi}^\pi \exp(a\cos(\theta)) \exp(ai\sin(\theta))d\theta = 2\pi
$$

$$
\int_{-\pi}^\pi \exp(a\cos(\theta)) (\cos(a\sin(\theta)) + i\sin(a\sin(\theta))) d\theta = 2\pi
$$

$$
\int_{-\pi}^\pi \exp(a\cos(\theta)) \cos(a\sin(\theta)) + i\exp(a\cos(\theta))\sin(a\sin(\theta)) d\theta = 2\pi
$$

The rhs has no imaginary component. So, the imaginary part of the lhs must be zero.

$$
\int_{-\pi}^\pi \exp(a\cos(\theta)) \cos(a\sin(\theta))d\theta = 2\pi
$$

$$
\int_{-\pi}^0 \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta
+ \int_0^\pi \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta
= 2\pi
$$

Consider the first term. Swap the direction of integration.

$$
\int_{-\pi}^0 \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta =
- \int_0^{-\pi} \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta
$$

Perform the change of variables $\theta \to -\theta$

$$
\int_{-\pi}^0 \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta =
- \int_0^{\pi} \exp(a\cos(-\theta)) \cos(a\sin(-\theta)) (-d\theta)
$$

Then, sine is odd.

$$
\int_{-\pi}^0 \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta =
\int_0^{\pi} \exp(a\cos(-\theta)) \cos(-a\sin(\theta)) d\theta
$$

Cosine is even,

$$
\int_{-\pi}^0 \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta =
\int_0^{\pi} \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta
$$

Swap this back in,

$$
2\int_0^\pi \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta
= 2\pi
$$

Then,

$$
\int_0^\pi \exp(a\cos(\theta)) \cos(a\sin(\theta)) d\theta
= \pi
$$
