**2. By finding an antiderivative, evaluate each of these integrals, where the path is any contour between the indicated limits of integration:**

**(a)**$\int_0^{1 + i} z^2 dz$



This function is a polynomial, so it's analytic on $\mathbb{C}$. So, it has an antiderivative. This is given by,

$$
\int z^2 dz = 3^{-1}z^3
$$

So, for any path between the bounds in $\mathbb{C}$,

$$
\int_0^{1 + i} z^2 dz = \bigg[3^{-1}z^3\bigg]_0^{1 + i}
$$

$$
\int_0^{1 + i} z^2 dz = \bigg[
    3^{-1}(1 + i)^3 -
    3^{-1}(0)^3
    \bigg]
$$

$$
\int_0^{1 + i} z^2 dz = 3^{-1}(1 + i)^3
$$

$$
\int_0^{1 + i} z^2 dz = 3^{-1}(\sqrt{2}\exp(i\pi/4))^3
$$

$$
\int_0^{1 + i} z^2 dz = 3^{-1}\sqrt{2}^3\exp(i3\pi/4)
$$

$$
\int_0^{1 + i} z^2 dz = {2 \over 3}\sqrt{2}\exp(i3\pi/4)
$$

$$
\int_0^{1 + i} z^2 dz = {2 \over 3}(-1 + i)
$$

**(b)** $\int_0^{\pi + 2i} \cos(z/2)dz$

Take the antiderivative,

$$
\int \cos(z/2) dz = 2\sin(z/2)
$$

The function $\cos(z/2)$ is analytic on $\mathbb{C}$. So, for any path between the bounds,

$$
\int_0^{\pi + 2i} \cos(z/2) dz = 2\sin((\pi + 2i)/2) - 2\sin((0)/2)
$$

$$
\int_0^{\pi + 2i} \cos(z/2) dz = 2\sin(\pi/2 + i)
$$

$$
\int_0^{\pi + 2i} \cos(z/2) dz = 2{e^{i(\pi/2 + i)} - e^{-i(\pi/2 + i)} \over 2i}
$$

$$
\int_0^{\pi + 2i} \cos(z/2) dz = -i(e^{i\pi/2 - 1} - e^{-i\pi/2 + 1})
$$

$$
\int_0^{\pi + 2i} \cos(z/2) dz = -i(e^{i\pi/2}e^{-1} - e^{-i\pi/2}e)
$$

$$
\int_0^{\pi + 2i} \cos(z/2) dz = -i(ie^{-1} - (-i)e)
$$

$$
\int_0^{\pi + 2i} \cos(z/2) dz = -i(ie^{-1} + ie)
$$

$$
\int_0^{\pi + 2i} \cos(z/2) dz = e^{-1} + e
$$
