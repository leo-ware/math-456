**2. $f (z)= z− 1$ and $C$ is the arc from $z = 0$ to $z = 2$ consisting of**

**(a) the semicircle $z = 1 + e^{iθ}$ $(π ≤ θ ≤ 2π)$**

Evaluate,

$$
dz = i\exp(i\theta)d\theta
$$

$$
f(z(\theta)) = \exp(i\theta)
$$

Then,

$$
\int_C f(z)dz = \int_0^{2\pi}(\exp(i\theta))(i\exp(i\theta)d\theta)
$$

$$
\int_C f(z)dz = \int_0^{2\pi}i\exp(2i\theta)d\theta
$$

$$
\int_C f(z)dz = i\left[ (2i)^{-1}\exp(2i\theta) \right]_0^{2\pi}
$$

$$
\int_C f(z)dz = i\left[ -{i \over 2}\exp(2i\theta) \right]_0^{2\pi}
$$

$$
\int_C f(z)dz = {1 \over 2}\bigg[ \exp(2i\theta) \bigg]_0^{2\pi}
$$

$$
\int_C f(z)dz = {1 \over 2}\bigg(
    \exp(2i(2\pi))
    - \exp(2i(0))
\bigg)
$$

$$
\int_C f(z)dz = {1 \over 2}\bigg(
    \exp(i4\pi)
    - \exp(0)
\bigg)
$$

$$
\int_C f(z)dz = {1 \over 2}\bigg(
    1 - 1
\bigg)
$$

$$
\int_C f(z)dz = 0
$$

**(b) the segment $z = x$ $(0 ≤ x ≤ 2)$ of the real axis.**

Evaluate,

$$
f(z(x)) = x - 1
$$

$$
dz = dx
$$

So,

$$
\int_C zdz = \int_0^2 (x - 1)dx

$$
\int_C zdz = \bigg[{1 \over 2}x^2 - x \bigg]_0^2
$$

$$

$$

$$
\int_C zdz = 0
