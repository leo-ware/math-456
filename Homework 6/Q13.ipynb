**1. Apply the Cauchy–Goursat theorem to show that**

$$
\int_C f(z)dz = 0
$$

**when the contour $C$ is the unit circle $|z| = 1$ in either direction and when**

**(a)** $f(z) = {z^2 \over z + 3}$

Consider the open region $D \subset \mathbb{C}$ defined by the set ${x : |x| \le 2}$. Note this set is simply connected. Note the contour $C$ is a simple, connected curve fully contained in $D$.

The two functions $z^2$ and $z + 3$ are polynomials, so they have derivatives. Additionally, the denominator $z + 3$ is nonzero on $D$. So, the quotient has a derivative on $D$. So, $f$ is analytic on $D$.

So, by the Cauchy-Goursat theorem, the integral is equal to zero.

**(c)** $f(z) = {1 \over z^2 + 2z + 2}$

Consider the denominator.

$$
{d \over dz}(z^2 + 2z + 2) = 2z + 2
$$

$$
{d^2 \over d^2z}(z^2 + 2z + 2) = 2
$$

Set to zero to find critical points $z^*$.

$$
2z^* + 2 = 0
$$

$$
z^* = -1
$$

At $z^* = -1$,

$$
z^2 + 2z + 2 = (-1)^2 + 2(-1) + 2 = 2
$$

So, the only optimum is when $z = -1$, at this point $f(z) = 2$. Since the second derivative is positive. So, this is a minimum. So, $f(z) \gt 0$.

Consider the open region $D \subset \mathbb{C}$ defined by the set ${x : |x| \le 2}$. Note this set is simply connected. Note the contour $C$ is a simple, connected curve fully contained in $D$.

The function $f$ is the ratio of a constant and a nonzero polynomial on $D$. So, it is differentiable on $D$. So, it's analytic on $D$.

So, by the Cauchy-Goursat theorem, the integral is equal to zero.

**(f)** $f(z) = \operatorname{Log}(z + 2)$

Take the derivative,

$$
{d \over dz}f(z) = {1 \over z + 2}
$$

This is defined whenever $z \ne -2$. This never occurs on $D$. So, $f$ is differentiable on $D$. So, $f$ is analytic on $D$. So, $f$ is analytic on $C$.

So, by the Cauchy-Goursat theorem, the integral is equal to zero.
