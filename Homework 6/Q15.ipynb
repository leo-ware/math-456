**6. Let $C$ denote the positively oriented boundary of the half disk $0 ≤ r ≤ 1$, $0 ≤ θ ≤ π$, and let $f (z)$ be a continuous function defined on that half disk by writing $f (0)= 0$ and using the branch**

$$
f(z) = \sqrt{r}\exp(i\theta/2) \quad (r \gt 0, -\pi/2 \lt \theta \lt 3\pi/2)
$$

**of the multiple valued function $z^{1/2}$. Show that,**

$$
\int_C f(z)dz = 0
$$

**by evaluating separately the integrals of $f(z)$ over the semicircle and the two radii which make up $C$. Why does the Cauchy–Goursat theorem not apply here?**

Cauchy-Goursat doesn't apply because the function isn't analytic at the origin. It's continuous after the patch $f(0) = 0$, but the derivative doesn't exist.

$$
f'(0) = \lim_{h \to 0} {f(h) - f(0) \over h - 0}
$$

$$
f'(0) = \lim_{h \to 0} {f(h) \over h}
$$

$$
f'(0) = \lim_{h \to 0} {\sqrt{r}\exp(i\theta/2) \over r\exp(i\theta)}
$$

$$
f'(0) = \lim_{h \to 0} {1 \over \sqrt{r}\exp(i\theta)}
$$

This goes to infinity as $r \to 0$.

We'll do the semicircle first.

Let,

$$
C_0 : z = \exp(i\theta) \quad \theta : 0 \to \pi
$$

Then,

$$
dz = i\exp(i\theta)d\theta
$$

$$
f(z) = \exp(i\theta/2)
$$


Then,

$$
\int_{C_0} f(z)dz = \int_0^\pi (\exp(i\theta/2))(i\exp(i\theta)d\theta)
$$

$$
\int_{C_0} f(z)dz = i\int_0^\pi \exp\left({3i\theta \over 2}\right)d\theta
$$

$$
\int_{C_0} f(z)dz = i\left[ \left({3i \over 2}\right)^{-1} \exp\left({3i\theta \over 2}\right) \right]_0^\pi
$$

$$
\int_{C_0} f(z)dz = i\left(-{2i \over 3}\right) \left[  \exp\left({3i\theta \over 2}\right) \right]_0^\pi
$$

$$
\int_{C_0} f(z)dz = {2 \over 3} \left( \exp\left({3i\pi \over 2}\right) - \exp\left({3i(0) \over 2}\right) \right)
$$

$$
\int_{C_0} f(z)dz = {2 \over 3} \left( -i - 1 \right)
$$

$$
\int_{C_0} f(z)dz = {2(-1 - i) \over 3}
$$

Let,

$$
C_1: z = r \quad r: 0 \to 1
$$

Then,

$$
dz = dr
$$

$$
f(z) = r^{1/2}
$$

Then,

$$
\int_{C_1} f(z)dz = \int_0^1 r^{1/2} dr
$$

$$
\int_{C_1} f(z)dz = \bigg[ (2/3)r^{3/2} \bigg]_0^1 = 2/3
$$

On this branch, $r \gt 0$. So, to represent the left radius, we will do,

Let,

$$
C_2: z = r\exp(i\pi) \quad r: 1 \to 0
$$

Then,

$$
dz = \exp(i\pi)dr = -dr
$$

$$
f(z) = r^{1/2}\exp(i\pi/2)
$$

Then,

$$
\int_{C_2} f(z)dz = \int_1^0 -r^{1/2}\exp(i\pi/2) dr
$$

$$
\int_{C_2} f(z)dz = \exp(i\pi/2) \int_0^1 r^{1/2} dr
$$

$$
\int_{C_2} f(z)dz = \exp(i\pi/2) \bigg[ (2/3)r^{3/2} \bigg]_0^1
$$

$$
\int_{C_2} f(z)dz = {2i\over 3}
$$

So,

$$
C = C_0 \cup C_1 \cup C_2
$$

$$
\int_C f(z) dz = {2(-1 - i) \over 3} + {2 \over 3} + {2i\over 3} = 0
$$
