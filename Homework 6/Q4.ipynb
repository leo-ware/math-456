**5. Suppose that a function $f (z)$ is analytic at a point $z_0 = z(t_0)$ lying on a smooth arc $z = z(t)$ $(a ≤ t ≤ b)$. Show that if $w(t)= f [z(t)]$, then**

$$
w'(t) = f'[z(t)]z'(t)
$$

**when t = $t_0$**

Break $f$ and $z$ into real and imaginary parts.

$$
f(z) = u(x, y) + iv(x, y)
$$

$$
z(t) = x(t) + iy(t)
$$

Take the derivative,

$$
{d \over dz} f(z) = u(x, y) + iv(x, y)
$$

$$
z(t) = x(t) + iy(t)
$$

We can rewrite $w$ as,

$$
w(t) = u(x(t), y(t)) + iv(x(t), y(t))
$$

Then,

$$
{d \over dt} w(t)
= {d \over dt} \bigg(u(x(t), y(t)) + iv(x(t), y(t))\bigg)
$$

$$
{d \over dt} w(t)
= {d \over dt} u(x(t), y(t)) + {d \over dt}iv(x(t), y(t))
$$

By the properties of complex derivatives,

$$
{d \over dt} w(t)
= {d \over dt} u(x(t), y(t)) + i{d \over dt}v(x(t), y(t))
$$

By the chain rule,

$$
{d \over dt} w(t)
= u_xx'(t) + u_yy'(t) + i(v_x x'(t) + v_y y'(t))
$$

Then,

$$
{d \over dt} w(t)
= (u_x + iv_x)x'(t) + (u_y + iv_y)y'(t)
$$

$$
{d \over dt} w(t)
= (u_x + iv_x)x'(t) + i(v_y - iu_y)y'(t)
$$

We have that $f$ is analytic, so,

$$
f' = u_x + iv_x = v_y - iu_y
$$

So,

$$
{d \over dt} w(t)
= f'x'(t) - if'y'(t)
$$

$$
{d \over dt} w(t)
= f'(x'(t) + iy'(t))
$$

We have that $z(t)$ is a smooth curve. So, $z(t)$ is differentiable. So,

$$
z' = x' + iy'
$$

So,

$$
{d \over dt} w(t)
= f'z'
$$
