Let $C_0$ denote the positively oriented circle centered at $2 + i$ with radius 10. By 46.13,

$$
\int_{C_0} (z - (2 + i))^{n - 1}dz =
\begin{cases}
0 & \text{when } n \in \mathbb{Z} / \{0\} \\
2\pi i & \text{when } n = 0
\end{cases}
$$

Note that $C_0$ contains $C$.

Consider the function,

$$
f(z) = (z - (2 + i))^{n - 1}
$$

Take the derivative.

$$
{d \over dz}f(z) = {d \over dz}(z - (2 + i))^{n - 1} = (n - 1)(z - (2 + i))^{n - 2}
$$

Note that the function and its derivative are defined when $z \ne 2 + i$. This point is interior to $C$. So, $f$ is analytic on $C$, $C_0$, and on points interior to $C_0$ but exterior to $C$. This suffices to appy the Cauchy_goursat theorem. Let $-C$ denote the points in $C$ oriented in the negative direction. Then,

$$
\int_{C_0} f(z)dz + \int_{-C} f(z)dz = 0
$$

Then,

$$
\int_{C_0} f(z)dz - \int_{C} f(z)dz = 0
$$

$$
\int_{C_0} f(z)dz = \int_{C} f(z)dz
$$

So,

$$
\int_{C} f(z)dz =
\begin{cases}
0 & \text{when } n \in \mathbb{Z} / \{0\} \\
2\pi i & \text{when } n = 0
\end{cases}
$$
