**b.** $\int_{1}^2 ({1 \over t} - i)^2 dt$

Let,

$$
x = \int_{1}^2 ({1 \over t} - i)^2 dt
$$

Then,

$$
x = \int_{1}^2 t^{-2} -{2i \over t} + 1 dt
$$

$$
x = \int_{1}^2 t^{-2} - {2i \over t} + 1 dt
$$



$$
x = \bigg[ -t^{-1} - 2i \operatorname{Log}(t) + t \bigg]_{t = 1}^2
$$

$$
x = \bigg[ -t^{-1} - 2i (\ln |t| + i\operatorname{Arg}(t)) + t \bigg]_{t = 1}^2
$$

$$
x = \bigg[ -t^{-1} - 2i \ln |t| + 2 \operatorname{Arg}(t) + t \bigg]_{t = 1}^2
$$

$$
x = \bigg[ (-t^{-1} + t + 2 \operatorname{Arg}(t)) -2 \ln |t|i \bigg]_{t = 1}^2
$$

Then,

$$
x =
\bigg( (-2^{-1} + 2 + 2 \operatorname{Arg}(2)) -2 \ln |2|i) \bigg) -
\bigg( (-1^{-1} + 1 + 2 \operatorname{Arg}(1)) -2 \ln |1|i \bigg)
$$

$$
x =
\bigg( -2^{-1} + 2 - 2 \ln |2|i \bigg) -
\bigg( -1^{-1} + 1 \bigg)
$$

$$
x =
-{1 \over 2} - 1 + 1 -2 \ln(2)i
$$

$$
x =
-{1 \over 2} + -\ln(4)i
$$

**d.** $\int_{0}^{\infty} e^{-zt} dt$ where $\operatorname{Re}(z) \gt 0$



$$
\int e^{-zt} dt = -z^{-1}e^{-zt}
$$

So,

$$
\int_0^\infty e^{-zt} dt = (\lim_{t \to \infty} [-z^{-1}e^{-zt}]) - (-z^{-1}e^{-z(0)})
$$

$$
\int_0^\infty e^{-zt} dt = z^{-1}
$$
