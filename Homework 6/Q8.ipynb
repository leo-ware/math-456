**9. Let $C$ denote the positively oriented unit circle $|z| = 1$ about the origin.**

**(a) Show that if $f(z)$ is the principal branch**

$$
z^{-3/4} = \exp(-{3 \over 4}\operatorname{Log}(z)) \quad (|z| \gt 0, -\pi \lt \operatorname{Arg}z \lt \pi)
$$

**of $z^{-3/4}$, then,**

$$
\int_C f(z)dz = 4\sqrt{2}i
$$

We can parametrize $C$ as,

$$
C: z = \exp(i\theta) \quad \theta: -\pi \to \pi
$$

Then,

$$
dz = i\exp(i\theta)d\theta
$$

Also,

$$
f(z(\theta)) = \exp \left(-{3 \over 4}\operatorname{Log}(\exp(i\theta)) \right)
$$

Then,

$$
f(z(\theta)) = \exp \left(-{3 \over 4}i\theta \right)
$$

Integrate,

$$
\int_C f(z)dz = \int_{-\pi}^\pi \exp \left(-{3 \over 4}i\theta \right)(i\exp(i\theta)) d\theta
$$

$$
\int_C f(z)dz = \int_{-\pi}^\pi
i \exp \left(-{3 \over 4}i\theta + i\theta \right)
d\theta
$$

$$
\int_C f(z)dz = \int_{-\pi}^\pi
i \exp \left({1 \over 4}i\theta \right)
d\theta
$$

$$
\int_C f(z)dz = 
i\left[
    ({1 \over 4}i)^{-1}\exp \left( {1 \over 4}i\theta \right)
\right]_{-\pi}^\pi
$$



$$
\int_C f(z)dz = 
i\left[
    (-4i)
    \exp \left( {1 \over 4}i\theta \right)
\right]_{-\pi}^\pi
$$

$$
\int_C f(z)dz = 
4\left(
    \exp \left( {1 \over 4}i\pi \right)
    - \exp \left( {1 \over 4}i(-\pi) \right)
\right)
$$

$$
\int_C f(z)dz = 
4{2i \over 2i}\left(
    \exp \left( {1 \over 4}i\pi \right)
    - \exp \left( -{1 \over 4}i\pi \right)
\right)
$$

$$
\int_C f(z)dz = 
8i {
    \exp \left( {1 \over 4}i\pi \right)
    - \exp \left( -{1 \over 4}i\pi \right)
    \over 2i
}
$$

$$
\int_C f(z)dz = 
8i \sin(\pi/4)
$$

$$
\int_C f(z)dz = 
{8i \over \sqrt{2}}
$$

$$
\int_C f(z)dz = 
4\sqrt{2}i
$$

**(b) Show that if $g(z)$ is the branch**

$$
z^{-3/4} = \exp(-{3 \over 4}\operatorname{Log}(z)) \quad (|z| \gt 0, 0 \lt \operatorname{arg}z \lt 2\pi)
$$

**of the same power function in part (a), then**

$$
\int_C g(z)dz = -4 + 4i
$$

**This exercise demonstrates how the value of an integral of a power function depends in general on the branch that is used.**

We will again parametrize $C$ as,

$$
C: \exp(i\theta) \quad \theta: 0 \to 2pi
$$

Then,

$$
dz = i\exp(i\theta)d\theta
$$

As demonstrated earlier,

$$
f(z(\theta)) = \exp \left( -{3 \over 4}i\theta \right)
$$

Integrate,

$$
\int_C f(z)dz = \int_0^{2\pi} \exp \left( -{3 \over 4}i\theta \right) (i\exp(i\theta)) d\theta
$$

$$
\int_C f(z)dz = \int_0^{2\pi} i \exp \left( -{3 \over 4}i\theta + i\theta \right) d\theta
$$

$$
\int_C f(z)dz = i \int_0^{2\pi} \exp \left( {1 \over 4}i\theta \right) d\theta
$$

$$
\int_C f(z)dz = i \left[ ({1 \over 4}i)^{-1} \exp \left( {1 \over 4}i\theta \right) \right]_0^{2\pi}
$$

$$
\int_C f(z)dz = i \left[
    -4i
    \exp \left( {1 \over 4}i\theta \right)
\right]_0^{2\pi}
$$

$$
\int_C f(z)dz = 4 \left[
    \exp \left( {1 \over 4}i(2\pi) \right) -
    \exp \left( {1 \over 4}i(0) \right)
\right]
$$

$$
\int_C f(z)dz = 4 \left[
    \exp \left( i\pi/2 \right) -
    \exp(0)
\right]
$$

$$
\int_C f(z)dz = 4 (i - 1)
$$

$$
\int_C f(z)dz = 4i - 4
$$
