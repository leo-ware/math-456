**4. According to definition (2), Sec. 42, of definite integrals of complex-valued functions of a real variable**

$$
\int_0^\pi e^{(1 + i)x} dx = \int_0^\pi e^x\cos(x)dx + i\int_0^\pi e^x\sin(x)dx
$$

**Evaluate the two integrals on the right here by evaluating the single integral on the left and then using the real and imaginary parts of the value found.**

$$
\int e^{(1 + i)x}dx = (1 + i)^{-1}e^{(1 + i)x}
$$

$$
\int e^{(1 + i)x}dx = {1 - i \over 2}e^{(1 + i)x}
$$

So,

$$
\int_0^\pi e^{(1 + i)x}dx = \left[{1 - i \over 2}e^{(1 + i)x}\right]_{x = 0}^\pi
$$

$$
\int_0^\pi e^{(1 + i)x}dx =
({1 - i \over 2}e^{(1 + i)\pi})
- ({1 - i \over 2}e^{(1 + i)(0)})
$$

$$
\int_0^\pi e^{(1 + i)x}dx =
{1 - i \over 2}e^{(1 + i)\pi}
- {1 - i \over 2}
$$

$$
\int_0^\pi e^{(1 + i)x}dx =
{1 - i \over 2}(e^{(1 + i)\pi} - 1)
$$

$$
\int_0^\pi e^{(1 + i)x}dx =
{1 - i \over 2}(e^{\pi}e^{i\pi} - 1)
$$

$$
\int_0^\pi e^{(1 + i)x}dx =
{1 - i \over 2}(-e^{\pi} - 1)
$$

$$
\int_0^\pi e^{(1 + i)x}dx =
{i - 1 \over 2}(e^{\pi} + 1)
$$

$$
\int_0^\pi e^{(1 + i)x}dx =
- {1 \over 2}(e^{\pi} + 1)
+ {i \over 2}(e^{\pi} + 1)
$$

So,

$$
\operatorname{Re} \int_0^\pi e^{(1 + i)x}dx =
- {1 \over 2}(e^{\pi} + 1)
$$

$$
\operatorname{Imag} \int_0^\pi e^{(1 + i)x}dx =
{1 \over 2}(e^{\pi} + 1)
$$

We have that,

$$
\int_0^\pi e^{(1 + i)x} dx = \int_0^\pi e^x\cos(x)dx + i\int_0^\pi e^x\sin(x)dx
$$

So,

$$
\operatorname{Re} \int_0^\pi e^{(1 + i)x} dx =
\int_0^\pi e^x\cos(x)dx
$$

$$
\operatorname{Imag} \int_0^\pi e^{(1 + i)x} dx =
\int_0^\pi e^x\sin(x)dx
$$


So,

$$
- {1 \over 2}(e^{\pi} + 1) = 
\int_0^\pi e^x\cos(x)dx
$$

$$
{1 \over 2}(e^{\pi} + 1) =
\int_0^\pi e^x\sin(x)dx
$$
