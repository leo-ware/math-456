**$f(z)$ is the principal branch $z^{a−1} = \exp[(a−1) \operatorname{Log} z]$ for $|z| \gt 0$ and $-\pi \lt \operatorname{Arg}z \lt \pi$ of the power function $z^{a−1}$, where $a$ is a nonzero real number, and $C$ is the positively oriented circle of radius $R$ about the origin.**

We can parametrize $C$ using the form,

$$
C: z = R\exp(i\theta) \quad \theta: -\pi \to \pi
$$

Then,

$$
dz = Ri\exp(i\theta)d\theta
$$

Also,

$$
f(z(\theta)) = \exp[(a−1) \operatorname{Log} (R\exp(i\theta))]
$$

We can simplify,

$$
\operatorname{Log} (R\exp(i\theta)) = \ln(R) + i\theta
$$

So,

$$
f(z(\theta)) = \exp[(a−1)(\ln(R) + i\theta)]
$$

$$
f(z(\theta)) = \exp[(a−1)\ln(R)]\exp[(a−1)i\theta]
$$

$$
f(z(\theta)) = R^{a - 1}\exp(i\theta(a−1))
$$

Integrate $f$ over $C$.

$$
\int_C f(z)dz = \int_{-\pi}^\pi (R^{a - 1}\exp(i\theta(a−1)))(Ri\exp(i\theta))d\theta
$$

$$
\int_C f(z)dz = \int_{-\pi}^\pi (iR^{a - 1}R\exp(i(\theta(a−1) + \theta)))d\theta
$$

$$
\int_C f(z)dz = \int_{-\pi}^\pi iR^a\exp(ia\theta)d\theta
$$

$$
\int_C f(z)dz = iR^a \bigg[ (ia)^{-1}\exp(ia\theta) \bigg]_{-\pi}^\pi
$$

$$
\int_C f(z)dz = iR^a(ia)^{-1} \bigg[ \exp(ia\pi) - \exp(-ia\pi) \bigg]
$$

$$
\int_C f(z)dz = {2i \over 2i}{R^a \over a} \bigg[ \exp(ia\pi) - \exp(-ia\pi) \bigg]
$$

$$
\int_C f(z)dz = 2i{R^a \over a} {\exp(ia\pi) - \exp(-ia\pi) \over 2i}
$$

$$
\int_C f(z)dz = 2i{R^a \over a} \sin(a\pi)
$$
