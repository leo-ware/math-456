**7. Show that if $C$ is a positively oriented simple closed contour, then the area of the region enclosed by $C$ can be written**

$$
{1 \over 2i}\int_C \bar{z}dz
$$

We have,

$$
\int_C f(z)dz = \int_C(udx - vdy) + i\int_C(udy + vdx)
$$

Consider the case when $f(z) = \bar{z}$.

Then,

$$
u = x
$$

$$
v = -y
$$

So,

$$
\int_C \bar{z}dz = \int_C(xdx - ydy) + i\int_C(xdy - ydx)
$$


Consider the real part. Consider some parametrization by $t: a \to b$.

$$
\operatorname{Re}\left[\int_C \bar{z}dz\right] = \int_a^b x(t)x'(t) - y(t)y'(t) dt
$$

Then,

$$
\operatorname{Re}\left[\int_C \bar{z}dz\right] = {1 \over 2} \left[x(t)^2 - y(t)^2\right]_a^b
$$

$$
\operatorname{Re}\left[\int_C \bar{z}dz\right] = {1 \over 2} \left[(x(b)^2 - y(b)^2) - (x(a)^2 - y(a)^2)\right]_a^b
$$

But, $C$ is a closed curve. So, $x(a) = x(b)$ and $y(a) = y(b)$.

$$
\operatorname{Re}\left[\int_C \bar{z}dz\right] = 0
$$


Consider the imaginary part.

$$
\operatorname{Imag}\left[\int_C \bar{z}dz\right] = \int_C(xdy - ydx)
$$

We can apply Green's theorem. Let $R$ be the region enclosed by $C$.

$$
\operatorname{Imag}\left[\int_C \bar{z}dz\right] = \int \int_R({\partial x \over \partial x} + {\partial y \over \partial y}) dA
$$

$$
\operatorname{Imag}\left[\int_C \bar{z}dz\right] = \int \int_R 2 dA
$$

Let $\phi$ denote the area of $R$.

$$
\operatorname{Imag}\left[\int_C \bar{z}dz\right] = 2\phi
$$

Recombine real and imaginary parts.

$$
\int_C \bar{z}dz = 0 + i2\phi
$$

Then,

$$
\phi = {1 \over 2i}\int_C \bar{z}dz
$$
