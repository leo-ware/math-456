**1. Let $C$ denote the positively oriented boundary of the square whose sides lie along the lines $x = ±2$ and $y = ±2$. Evaluate each of these integrals:**

**(a)** $\int_C {e^{-z} \over z - i\pi/2}dz$

Let,

$$
g(z) = e^{-z}
$$

Then, $g$ is analytic on $\mathbb{C}$.

Let $z_0 = i\pi/2$. The integral can be rewritten,

$$
\int_C f(z)dz = \int_C {g(z) \over z - z_0}
$$

Then, since this point is interior to the contour, by the Cauchy integral theorem.

$$
\int_C f(z)dz = 2\pi i f(z_0)
$$

$$
\int_C f(z)dz = 2\pi i \exp(-i\pi/2)
$$

$$
\int_C f(z)dz = 2\pi i (-i)
$$

$$
\int_C f(z)dz = 2\pi
$$

**(c)** $\int_C {z \over 2z + 1} dz$

The integral can be rewritten,

$$
\int_C {z \over 2z + 1} = \int_C {z/2 \over z - (-1/2)}
$$

The numerator $z/2$ is analytic on $\mathbb{C}$, and the point -1/2 is interior to the contour. So, by the Cauchy integral theorem, this can be rewritten as,

$$
\int_C {z \over 2z + 1} = 2\pi i ((-1/2)/2)
$$

So,

$$
\int_C {z \over 2z + 1} = -i \pi /2
$$
