**5. Show that**

$$
\int_{-1}^1 z^i dz = {1 + e^{-\pi} \over 2}(1 - i)
$$

**where the integrand denotes the principal branch**

$$
z^i = \exp(i\operatorname{Log}z) \quad (|z| \gt 0, -\pi \lt \operatorname{Arg}z \lt \pi)
$$

**of $z^i$ and where the path of integration is any contour from $z = −1$ to $z = 1$ that, except for its end points, lies above the real axis. (Compare with Exercise 6, Sec. 46.)**

We will consider another branch,

$$
z^i = \exp(i\operatorname{log} z) \quad (|z| \gt 0, -\pi/2 \lt \operatorname{arg}z \lt 3\pi/2)
$$

Note that, at every point that lies above the real axis, the two branches coincide. So, integration of the prior branch is equivalent to integration of this branch. Additionallty, along paths that stay above the real axis, we never cross the branch cut.

We take the antiderivative,

$$
\int z^i dz = (1 + i)^{-1}z^{1 + i}
$$

$$
\int z^i dz = {1 - i \over 2}z^{1 + i}
$$

Then,

$$
\int_{-1}^1 z^i dz = {1 - i \over 2}\bigg[z^{1 + i}\bigg]_{z = -1}^1
$$

$$
\int_{-1}^1 z^i dz = {1 - i \over 2}\bigg[\exp((1 + i)\log(z))\bigg]_{z = -1}^1
$$

$$
\int_{-1}^1 z^i dz = {1 - i \over 2}
\bigg(
    \exp((1 + i)\log(1))
    - \exp((1 + i)\log(-1))
\bigg)
$$

$$
\int_{-1}^1 z^i dz = {1 - i \over 2}
\bigg(
    \exp((1 + i)(0))
    - \exp((1 + i)(i\pi))
\bigg)
$$

$$
\int_{-1}^1 z^i dz = {1 - i \over 2}
(
    1
    - \exp(i\pi - \pi)
)
$$

$$
\int_{-1}^1 z^i dz = {1 - i \over 2} (1 - \exp(i\pi)\exp(- \pi))
$$

$$
\int_{-1}^1 z^i dz = {1 - i \over 2} (1 + \exp(- \pi))
$$
