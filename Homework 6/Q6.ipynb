**5. $f (z)= 1$ and $C$ is an arbitrary contour from any fixed point $z_1$ to any fixed point $z_2$ in the $z$ plane.**

Observe that $f$ is analytic on the complex plane. So, the integral from $z_1$ to $z_2$ is path independent. So, we can integrate along the line,

$$
C : z = (1 - t)z_1 + tz_2 \quad t : 0 \to 1
$$

Rewrite $z$ as,

$$
z = z_1 + t(z_2 - z_1)
$$

Then,

$$
{dz \over dt} = \left( {dz \over dt}z_1 + t(z_2 - z_1) \right) = z_2 - z_1
$$

$$
dz = (z_2 - z_1)dt
$$

Integrate over $C$.

$$
\int_C f(z) dz = \int_0^1 f(z(t))(z_2 - z_1)dt
$$

$$
\int_C f(z) dz = \int_0^1 (z_2 - z_1)dt
$$

$$
\int_C f(z) dz = \bigg[ (z_2 - z_1)t \bigg]_0^1
$$

$$
\int_C f(z) dz = z_2 - z_1
$$
