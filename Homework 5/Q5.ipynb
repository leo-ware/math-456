**1. Show that**

**(a)** $\operatorname{Log}(-ei) = 1 - {\pi \over 2}i$

Note that $-i = e^{-i\pi/2}$. So,
$$
-ei = ee^{-i\pi/2}
$$

So, $-ei$ has mod $e$ and angle $-\pi/2$. So,

$$
\operatorname{Log}(-ei) = \ln e - i{\pi \over 2}
$$

Then,

$$
\operatorname{Log}(-ei) = 1 - i{\pi \over 2}
$$

**(b)** $\operatorname{Log}(1-i) = {1 \over 2}\ln 2 - {\pi \over 4}i$

Note that,

$$
1 - i = \sqrt{2}e^{-{\pi \over4}i}
$$

So, $1 - i$ has mod $\sqrt{2}$ and angle $-\pi/4$. So,

$$
\operatorname{Log}(1-i) = \ln \sqrt{2} - i{\pi \over 4}
$$

Then,

$$
\operatorname{Log}(1-i) = {1 \over 2}\ln 2 - {\pi \over 4}i
$$
