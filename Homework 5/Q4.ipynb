**11. Describe the behavior of $e^z = e^x e^{iy}$ as (a) $x$ tends to $−∞$; (b) $y$ tends to $∞$.**

We have that,

$$
e^z = e^xe^{yi}
$$

Then,

$$
|e^z| = e^x
$$

As $x \to -\infty$, $e^x \to 0$. So, $z$ is going to zero exponentially fast.

For $n \in \mathbb{Z}$,

$$
\operatorname{arg}(e^z) = y + 2\pi n
$$

As $y \to \infty$, the direction in which $e^z$ points is spinning around the origin counterclockwise at constant speed. There is no limit. It just does this forever.
