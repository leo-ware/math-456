**16. By using one of the identities (9) and (10) in Sec. 39 and then proceeding as in Exercise 15, Sec. 38, find all roots of the equation**

**(a)** $\sinh z = i$

Ans:
$$
z = (2n + 1/2)\pi i
$$

It is stipulated that,

$$
\sinh z = i
$$

Let $x$ and $y$ be real numbers such that $z = x + yi$. Applying equation 9, we derive,

$$
\sinh(x) \cos(y) + i \cosh(x)\sin(y) = i
$$

Since $x$ and $y$ are real, $\sinh(x) \cos(y)$ and $\cosh(x)\sin(y)$ are real. So,

Take the real part.

$$
\operatorname{Re}(\sinh(x) \cos(y) + i \cosh(x)\sin(y)) = \operatorname{Re}(i)
$$

$$
\sinh(x) \cos(y) = 0
$$

One of them has to be zero. It is known that the zeros of $\sinh$ lie only along the imaginary axis, but $x$ is real. So, if $\sinh(x) = 0$, then $x = 0$. The zeroes of $\cos(y)$ are achieved when $y \in \{({1 \over 2} + n)\pi : n \in \mathbb{Z}\}$.

So, either $x = 0$ or $y \in \{({1 \over 2} + n)\pi : n \in \mathbb{Z}\}$.

Take the imaginary part.

$$
\operatorname{Imag}(\sinh(x) \cos(y) + i \cosh(x)\sin(y)) = \operatorname{Imag}(i)
$$

$$
\cosh(x)\sin(y) = 1
$$

We know from the real constraint that $x = 0$ or $y \in \{({1 \over 2} + n)\pi : n \in \mathbb{Z}\}$. We will consider what can be learned from the imaginary constraint. 

- *Case 1*: If $x = 0$, then $\cosh(x) = 1$. So, then $\sin(y) = 1$, and $y \in \{({1 \over 2} + n)\pi : n \in \mathbb{Z}\}$. So, in this case, $x = 0$ and $y \in \{({1 \over 2} + n)\pi : n \in \mathbb{Z}\}$.

- *Case 2*: If $y \in \{({1 \over 2} + n)\pi : n \in \mathbb{Z}\}$, then $\sin(y) = 1$. In this case, $\cosh(x)$ must equal 1, but this only occurs when $x = 0$. So, in this case, $x = 0$ and $y \in \{({1 \over 2} + n)\pi : n \in \mathbb{Z}\}$.

So, in either case, $x = 0$ and $y \in \{({1 \over 2} + n)\pi : n \in \mathbb{Z}\}$. So, we can conclude $x = 0$ and $y \in \{({1 \over 2} + n)\pi : n \in \mathbb{Z}\}$. This is the set of roots of the equation.

$$
z = ({1 \over 2} + n)i\pi: n \in \mathbb{Z}
$$


By the definition of $\cosh$ and $\sin$, we get,

$$
{e^x + e^{-x} \over 2}{e^{iy} - e^{-iy} \over 2i} = 1
$$

$$
{e^xe^{iy} + e^{-x}e^{iy} + e^x(- e^{-iy}) + e^{-x}(- e^{-iy}) \over 4i} = 1
$$

$$
{e^{x + yi} + e^{-x +iy} - e^{x-iy} - e^{-x-yi} \over 4i} = 1
$$

$$
{e^{z} + e^{-\bar{z}} - e^{\bar{z}} - e^{-z} \over 4i} = 1
$$

$$
{e^{z} - e^{-z} \over 4i} + {e^{-\bar{z}} - e^{\bar{z}} \over 4i} = 1
$$

$$
\sin(z) + {e^{-\bar{z}} - e^{\bar{z}} \over 2i} = -i
$$

**(b)** $\cosh z = {1 \over 2}$

Ans:
$$
(2n \pm 1/3)\pi i
$$

We have that,

$$
\cosh z = {1 \over 2}
$$

Let $x$ and $y$ be real numbers such that $z = x + yi$. From theorem 10,

$$
\cosh x\cos y + i\sinh x \sin y = {1 \over 2}
$$

Take the imaginary part,

$$
\operatorname{Imag}(\cosh x\cos y + i\sinh x \sin y) = \operatorname{Imag}({1 \over 2})
$$

$$
\sinh x \sin y = 0
$$

So, either $\sinh x = 0$ or $\sin y = 0$. So, either $x = 0$ or $y = m\pi (m \in \mathbb{Z})$.

Take the real part,

$$
\operatorname{Re}(\cosh x\cos y + i\sinh x \sin y) = \operatorname{Re}({1 \over 2})
$$

$$
\cosh x\cos y = {1 \over 2}
$$

If $y = m\pi (m \in \mathbb{Z})$, $\cos y = \pm 1$. So, $|\cosh x|$ would have to equal ${1 \over 2}$, but $\cosh x$ is never less than 1. So, not $y = m\pi (m \in \mathbb{Z})$.

If $x = 0$, then $\cosh x = 1$. So, $\cos y = {1 \over 2}$.

$$
y = (2n \pm 1/3)\pi
$$

So,

$$
z = (2n \pm 1/3)\pi i
$$
