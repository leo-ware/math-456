**2. Show that**

**(b)** $\log i = \left( 2n + {1 \over 2} \right) \pi i$ for $n \in \mathbb{Z}$

Note that,

$$
i = e^{i\pi/2}
$$

So, $i$ has mod 1 and angle $\pi/2$. So,

$$
\operatorname{Log} i = \ln 1 + i{\pi \over 2}
$$

$$
\operatorname{Log} i = {1 \over 2}i\pi
$$

Then, for $n \in \mathbb{Z}$,

$$
\log i = {1 \over 2}i\pi + 2ni\pi
$$

Then,

$$
\log i = ({1 \over 2} + 2n)i\pi
$$
