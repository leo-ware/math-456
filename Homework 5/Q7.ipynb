**5. (a) Show that the two square roots of $i$ are $e^{i\pi / 4}$ and $e^{i5\pi / 4}$. Then show that,**

$$
\log(e^{i\pi / 4}) = \left( 2n + {1 \over 4} \right)\pi i \text{ for } n \in \mathbb{Z}
$$

**and**

$$
\log(e^{i\pi / 4}) = \left( (2n + 1) + {1 \over 4} \right)\pi i \text{ for } n \in \mathbb{Z}
$$

**Conclude that**

$$
\log(i^{1/2}) = \left( n + {1 \over 4} \right)\pi i \text{ for } n \in \mathbb{Z}
$$

Note that, for any $n \in \mathbb{Z}$,

$$
i = e^{i\pi(1/2 + 2n)}
$$

So,

$$
\sqrt{i} = (e^{i\pi(1/2 + 2n)})^{1/2}
$$

$$
\sqrt{i} = e^{i\pi(1/2 + 2n)(1/2)}
$$

$$
\sqrt{i} = e^{i\pi(1/4 + n)}
$$

Since this is a square root, there are two distinct roots, which occur when $n = 0$ and $n = 1$. So, the two square roots of $i$ are $e^{i\pi/4}$ and $e^{i5\pi/4}$.

Note that both of these roots have mod 1. The Arg of $e^{i\pi/4}$ is $\pi/4$ and the angle of $e^{i5\pi/4}$ is $5\pi/4$. So,

$$
\operatorname{Log}(e^{i\pi/4}) = \ln 1 + i{\pi \over 4}
$$

$$
\operatorname{Log}(e^{i\pi/4}) = i{\pi \over 4}
$$

Also,

$$
\operatorname{Log}(e^{i5\pi/4}) = \ln 1 + i{5\pi \over 4}
$$

$$
\operatorname{Log}(e^{i5\pi/4}) = i{5\pi \over 4}
$$

Then, for any $n \in \mathbb{Z}$,

$$
\log(e^{i\pi/4}) = i{\pi \over 4} + 2i\pi n
$$

$$
\log(e^{i\pi/4}) = (2 n + {1 \over 4})i\pi
$$

Also,

$$
\log(e^{5i\pi/4}) = i{5\pi \over 4} + 2i\pi n
$$

$$
\log(e^{5i\pi/4}) = ({5 \over 4} + 2 n)i\pi
$$

$$
\log(e^{5i\pi/4}) = ({1 \over 4} + 1 + 2 n)i\pi
$$

$$
\log(e^{5i\pi/4}) = ((2n + 1) + {1 \over 4})i\pi
$$

Consider the term $\log(i^{1/2})$. We have that $i^{1/2}$ is $e^{i\pi/4}$ and $e^{i5\pi/4}$. We have that,

$$
\log(e^{i\pi/4}) = (2 n + {1 \over 4})i\pi
$$

And,

$$
\log(e^{5i\pi/4}) = ((2n + 1) + {1 \over 4})i\pi
$$

So, $\log(i^{1/2})$ is all the values on the rhs of these two equations. Observe that this is exactly the set, $\{(n + {1 \over 4})i\pi : n \in \mathbb{Z}\}$, since $2n$ enumerates the even integers, and $2n + 1$ enumerates the odd integers, and the union of the odd and even integers is the integers.

So, for $n \in \mathbb{Z}$,

$$
\log(i^{1/2}) = (n + {1 \over 4})i\pi
$$

**(b) Show that**

$$
\log(i^{1/2}) = {1 \over 2} \log(i)
$$

**as stated in Example 5, Sec. 32, by finding the values on the right-hand side of this
equation and then comparing them with the final result in part (a).**

We can rewrite $i$ as,

$$
i = e^{i\pi/2}
$$

So, $|i| = 1$ and $\operatorname{Arg} i = \pi/2$. So,

$$
\operatorname{Log} i = \ln(1) + i\pi/2 = i\pi/2
$$

Then, for $n \in \mathbb{Z}$,

$$
\operatorname{log} i =  i\pi/2 + 2i\pi n = i\pi(1/2 + 2n)
$$

Then,

$$
{1 \over 2}\operatorname{log} i = {1 \over 2}i\pi(1/2 + 2n) = i\pi(1/4 + n)
$$

Recall that,

$$
\log{i^{1/2}} = i\pi(1/4 + n)
$$

So,

$$
\log{i^{1/2}} = {1\over 2}\log(i)
$$
