**7. Prove that $|\exp(−2z)| < 1$ if and only if $\operatorname{Re} z > 0$.**

Let $x$ and $y$ be real numbers such that,

$$
z = x + yi
$$

$$
-2z = -2x -2yi
$$

$$
\exp(-2z) = \exp(-2x -2yi)
$$

$$
\exp(-2z) = \exp(-2x)\exp(-2yi)
$$

$$
|\exp(-2z)| = |\exp(-2x)\exp(-2yi)|
$$

$$
|\exp(-2z)| = \exp(-2x)
$$

*Part 1:*

Suppose,

$$
\operatorname{Re} z \gt 0
$$

Then,

$$
x \gt 0
$$

$$
-2x \lt 0
$$

$$
\exp(-2x) \lt \exp(0)
$$

$$
\exp(-2x) \lt 1
$$

$$
|\exp(-2z)| \lt 1
$$

So,

$$
\operatorname{Re} z \gt 0 \implies|\exp(-2z)| \lt 1
$$

*Part 2:*

Suppose,

$$
|\exp(-2z)| \lt 1
$$

$$
\exp(-2x) \lt 1
$$

$$
-2x \lt \ln 1
$$

$$
-2x \lt 0
$$

$$
x \gt 0
$$

$$
\operatorname{Re} z \gt 0
$$

So,

$$
|\exp(-2z)| \lt 1 \implies \operatorname{Re} z \gt 0
$$

*Conclusion:*

We have that,

$$
\operatorname{Re} z \gt 0 \implies|\exp(-2z)| \lt 1
$$

$$
|\exp(-2z)| \lt 1 \implies \operatorname{Re} z \gt 0
$$

Therefore,

$$
\operatorname{Re} z \gt 0 \equiv |\exp(-2z)| \lt 1
$$
