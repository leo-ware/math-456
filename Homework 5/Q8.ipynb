Pages 95 & 96: problem 7

**7. Show that a branch (Sec. 33) $\log z = \ln r + i\theta$ $(r \gt 0, \alpha \lt \theta \lt 2\pi + \alpha)$ of the logarithmic function can we written,**

$$
\log z = {1 \over 2}\ln (x^2 + y^2) + i \tan^{-1}(y/x)
$$

**in reactangular coordinates. Then, using the theorem in Sec. 23, show that the given branch is analytic in its domain of definition and that**

$$
{d \over dz}\log z = {1 \over z}
$$

**there.**

It is necessary to specify what is meant by $\tan^{-1}(y/x)$. As typically understood, this expression is not defined for $x = 0$. So, there is no domain $r \gt 0, \alpha \lt \theta \lt 2\pi + \alpha$ on which this expression is defined. We will instead understand this expression as $\tan^{-1}(y/x) = \operatorname{Arg}(x + yi)$, where $\operatorname{Arg}(x + yi)$ is the unique branch of $\operatorname{arg}(x + yi)$ which has the range $(-\pi, \pi]$.

Let $z \ne 0$ be a complex number. Then, let $r = |z|$ and $\theta = \operatorname{Arg}z$. Let $x$ and $y$ be real numbers such that $z = x + yi$.

Then,

$$
r = (x^2 + y^2)^{1/2}
$$

By the convention stated above.

$$
\theta = \tan^{-1}(y/x)
$$

Consider the branch of the complex logarithm given by,

$$
\log(z) = \ln r + i\theta
$$

where $r \gt 0$ and $-\pi \lt \theta \lt \pi$.

We can substitute the above equations for $r$ and $\theta$.

$$
\log(z) = \ln ((x^2 + y^2)^{1/2}) + i\tan^{-1}(y/x)
$$

$$
\log(z) = {1 \over 2}\ln(x^2 + y^2) + i\tan^{-1}(y/x)
$$

Let $u$ and $v$ be real-valued functions of $x$, $y$ such that $\log(z) = u + iv$. Then,

$$
u = {1 \over 2}\ln(x^2 + y^2)
$$

$$
v = \tan^{-1}(y/x)
$$

Take the partials.

$$
u_x = {x \over x^2 + y^2}
$$

$$
u_y = {y \over x^2 + y^2}
$$

$$
v_x = {1 \over 1 + (y/x)^2}(-yx^{-2})
$$

$$
v_x = -{y \over (1 + (y/x)^2)x^2}
$$

$$
v_x = -{y \over x^2 + y^2}
$$

$$
v_y = {1 \over 1 + (y/x)^2}(x^{-1})
$$

$$
v_y = {1 \over (1 + (y/x)^2)x}
$$

$$
v_y = {x \over x}{1 \over (1 + (y/x)^2)x}
$$

$$
v_y = {x \over x^2 + y^2}
$$

We have that $x \ne 0$. So, the denominator $x^2 + y^2 \ne 0$. So, all of the partials are ratios of two nonzero continuous functions and are continuous. By inspection,

$$
u_x = v_y
$$

$$
u_y = -v_x
$$

So, the Cauchy-Riemann equations hold. So, the derivative exists and is given by,

$$
{d \over dx} \log z = u_x + v_x i
$$

Then,

$$
{d \over dx} \log z = {x \over x^2 + y^2} - {y \over x^2 + y^2} i
$$

$$
{d \over dx} \log z = {x - yi \over x^2 + y^2}
$$

$$
{d \over dx} \log z = {\overline{z} \over |z|^2}
$$

$$
{d \over dx} \log z = {\overline{z} \over z\overline{z}}
$$

Then,

$$
{d \over dx} \log z = {1 \over z}
$$

Since the derivative is defined everywhere, this function is analytic on the complex plane. So, it is analytic on the domain $r \gt 0$ and $-\pi \lt \theta \lt \pi$.
