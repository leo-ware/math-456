**Write $|\exp(2z + i)|$ and $|\exp(iz^2)|$ in terms of $x$ and $y$. Then show that $|\exp(2z + i) + \exp(iz^2)| \le e^{2x} + e^{-2xy}$.**

Let $x$ and $y$ be real numbers such that,

$$
z = x + yi
$$

Then,

$$
2z + i = 2x + (2y + 1)i
$$

$$
\exp(2z + i) = \exp(2x + (2y + 1)i)
$$

$$
\exp(2z + i) = \exp(2x) \exp((2y + 1)i)
$$

$$
|\exp(2z + i)| = |\exp(2x)|
$$

Since $x$ is real,

$$
|\exp(2z + i)| = \exp(2x)
$$

We have that,

$$
z = x + yi
$$

Then,

$$
z^2 = (x + yi)^2
$$

$$
z^2 = (x^2 - y^2) + 2xyi
$$

$$
iz^2 = (x^2 - y^2)i - 2xy
$$

$$
\exp(iz^2) = \exp((x^2 - y^2)i - 2xy)
$$

$$
\exp(iz^2) = \exp(- 2xy)\exp((x^2 - y^2)i)
$$

$$
|\exp(iz^2)| = |\exp(- 2xy)|
$$

Since $-2xy$ is real,

$$
|\exp(iz^2)| = \exp(-2xy)
$$

By the triangle inequality,

$$
|\exp(2z + i) + \exp(iz^2)| \le |\exp(2z + i)| + |\exp(iz^2)|
$$

Substituting the above expressions, we get,

$$
|\exp(2z + i) + \exp(iz^2)| \le e^{2x} + e^{-2xy}
$$
