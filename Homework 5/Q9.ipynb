**8. Find all roots of the equation $\log z = iπ/2$. (Ans. $z = i$)**

Let, $r = |z|$ and $\Theta = \operatorname{Arg} z$. By the definition of $\log$, we get that if $z \in \mathbb{Z}$, $\log z$ can take on any of the values,

$$
\log z = \ln r + (\Theta + 2n)i\pi
$$

Apply the given equation.

$$
\ln r + (\Theta + 2n)i\pi = i\pi/2
$$

Take the real part of both sides.

$$
\operatorname{Re}(\ln r + (\Theta + 2n)i\pi) = \operatorname{Re}(i\pi/2)
$$

Then,

$$
\ln r = 0
$$

$$
r = 1
$$

Take the imaginary part of both sides.

$$
\operatorname{Im}(\ln r + (\Theta + 2n)i\pi) = \operatorname{Im}(i\pi/2)
$$

Then,

$$
(\Theta + 2n)\pi = \pi/2
$$

$$
\Theta = 1/2 - 2n
$$

Using the results for length and Arg, we can rewrite $z$ as,

$$
z = e^{i\pi (1/2 - 2n)}
$$

$$
z = e^{i\pi/2}e^{-i\pi 2n}
$$

When $n \in \mathbb{Z}$, $e^{-i\pi 2n} = 1$. So,

$$
z = i
$$
