**10. (a) Show that if $e^z$ is real, then $\operatorname{Im} z = nπ$ where $(n = 0, ±1, ±2, . . .)$**

Let $x$ and $y$ be real such that,

$$
z = x + yi
$$

Then,

$$
e^z = e^{x + yi}
$$

$$
e^z = e^{x}e^{yi}
$$

$$
e^z = e^{x}(\cos(y) + i \sin(y))
$$

$$
e^z = e^{x}\cos(y) + i e^{x}\sin(y)
$$

$$
\operatorname{Imag}(e^z) = e^{x}\sin(y)
$$

Suppose $\operatorname{Imag}(e^z) = 0$, then,

$$
e^{x}\sin(y) = 0
$$

Since $x$ is real, $e^{x} \ne 0$. So,

$$
\sin(y) = 0
$$

So,

$$
y \in \{\pi n : n \in \mathbb{Z}\}
$$

But, $\operatorname{Imag}(z) = y$. So,

$$
\operatorname{Imag}(z) \in \{\pi n : n \in \mathbb{Z}\}
$$

So, 

$$
\operatorname{Imag}(e^z) = 0 \implies \operatorname{Imag}(z) \in \{\pi n : n \in \mathbb{Z}\}
$$

**(b) If $e^z$ is pure imaginary, what restriction is placed on $z$?**

We have that,

$$
e^z = e^{x}\cos(y) + i e^{x}\sin(y)
$$

Then,

$$
\operatorname{Re}(e^z) = \operatorname{Re}(e^{x}\cos(y) + i e^{x}\sin(y))
$$

$$
\operatorname{Re}(e^z) = e^{x}\cos(y)
$$

Suppose, $\operatorname{Re}(e^z) = 0$. Then,

$$
e^{x}\cos(y) = 0
$$

But, $x$ is real. So, $e^x \ne 0$. So,

$$
\cos(y) = 0
$$

So,

$$
y \in \{(n + {1 \over 2})\pi : n \in \mathbb{Z}\}
$$

But, $y = \operatorname{Imag}z$. So,

$$
\operatorname{Imag}z \in \{(n + {1 \over 2})\pi : n \in \mathbb{Z}\}
$$
